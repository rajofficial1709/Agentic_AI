**Import Libraries**

```python
from langchain.agents import initialize_agent, Tool
from langchain.agents import AgentType
from langchain.tools import WikipediaQueryRun
from langchain.utilities import WikipediaAPIWrapper
from langchain.tools.tavily_search import TavilySearchResults

from langchain.llms import HuggingFacePipeline
from transformers import pipeline
```

```python
import os
from getpass import getpass

os.environ["TAVILY_API_KEY"] = getpass("Enter Tavily API Key: ")
```

**FREE LLM**

```python
pipe = pipeline(
    "text-generation",
    model="gpt2",
    max_new_tokens=512
)

llm = HuggingFacePipeline(pipeline=pipe)
```

**Create Tools**

```python
# Web Search Tool
search = TavilySearchResults()

# Wikipedia Tool
wiki = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper())
```

**Initialize LLM**

**Creating Tool List**

```python
tools = [
    Tool(
        name="Web Search",
        func=search.run,
        description="Search latest info from the internet"
    ),
    Tool(
        name="Wikipedia",
        func=wiki.run,
        description="Get factual info from Wikipedia"
    )
]
```

**Creating Agent**

```python
agent = initialize_agent(
    tools,
    llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=False
)
```

**Report Function**

```python
def generate_report(topic):
    intro = llm.invoke(f"Explain {topic} in 4-5 simple lines.")

    findings = llm.invoke(f"What are the key benefits or findings of {topic}? Write in 4-5 lines.")

    challenges = llm.invoke(f"What are the challenges of {topic}? Write in 3-4 lines.")

    future = llm.invoke(f"What is the future scope of {topic}? Write in 3-4 lines.")

    conclusion = llm.invoke(f"Write a short conclusion about {topic} in 3-4 lines.")

    report = f"""
==============================
        COVER PAGE
==============================
Title: {topic}
Name: Your Name
Course: Your Course

==============================
1. Introduction
==============================
{intro}

==============================
2. Key Findings
==============================
{findings}

==============================
3. Challenges
==============================
{challenges}

==============================
4. Future Scope
==============================
{future}

==============================
5. Conclusion
==============================
{conclusion}
"""
    return report
```

** Agent Testing**

```python
topic = "Impact of AI in Healthcare"
report = generate_report(topic)

print(report)
```

```python
print(generate_report("Impact of AI in Healthcare"))
print(generate_report("Role of AI in Education"))
```
